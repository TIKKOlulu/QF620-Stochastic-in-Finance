1.Black-Scholes Model

Formulas：
$
d_1 = \frac{\ln\left(\frac{S}{K}\right) + \left(r + \frac{\sigma^2}{2}\right)T}{\sigma \sqrt{T}}, \quad d_2 = d_1 - \sigma \sqrt{T}
$


Vanilla Call Option Price:
$
C = S N(d_1) - K e^{-rT} N(d_2)
$


Vanilla Put Option Price:
$
P = K e^{-rT} N(-d_2) - S N(-d_1)
$


Digital Cash-or-Nothing Call Option Price:
$
C_{\text{digital}} = Q e^{-rT} N(d_2)
$


Digital Cash-or-Nothing Put Option Price:
$
P_{\text{digital}} = Q e^{-rT} N(-d_2)
$



Digital Asset-or-Nothing Call Option Price:（under risk-neutral valuation）
$
C_{\text{asset}} = S N(d_1)
$

Digital Asset-or-Nothing Put Option Price:（under risk-neutral valuation）
$
P_{\text{asset}} = S N(-d_1)
$



```python

def black_scholes_option(S, K, T, r, sigma, option_type='call'):
    d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    if option_type == 'call':
        price = S * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)
    elif option_type == 'put':
        price = K * np.exp(-r * T) * norm.cdf(-d2) - S * norm.cdf(-d1)
    return price

def black_scholes_digital_cash(S, K, T, r, sigma, Q=1, option_type='call'):
    d2 = (np.log(S / K) + (r - 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    if option_type == 'call':
        price = Q * np.exp(-r * T) * norm.cdf(d2)
    elif option_type == 'put':
        price = Q * np.exp(-r * T) * norm.cdf(-d2)
    return price

def black_scholes_digital_asset(S, K, T, r, sigma, option_type='call'):
    d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    if option_type == 'call':
        price = S * norm.cdf(d1)
    elif option_type == 'put':
        price = S * norm.cdf(-d1)
    return price
```

2.Bachelier Model

Formulas：
$
d = \frac{S - K}{\sigma \sqrt{T}}
$


Vanilla Call Option Price:
$
C = (S - K) N(d) + \sigma \sqrt{T} n(d)
$


Vanilla Put Option Price:
$
P = (K - S) N(-d) + \sigma \sqrt{T} n(d)
$

Digital Cash-or-Nothing Call Option Price:
$
C_{\text{digital}} = Q N\left(\frac{S - K}{\sigma \sqrt{T}}\right)
$


Digital Cash-or-Nothing Put Option Price:
$
P_{\text{digital}} = Q N\left(\frac{K - S}{\sigma \sqrt{T}}\right)
$



Digital Asset-or-Nothing Call Option Price:（under risk-neutral valuation）
$
C_{\text{asset}} = S N(d)
$

Digital Asset-or-Nothing Put Option Price:（under risk-neutral valuation）
$
P_{\text{asset}} = S N(-d)
$



```python
def bachelier_option(S, K, T, r, sigma, option_type='call'):
    d = (S - K) / (sigma * np.sqrt(T))
    discount = np.exp(-r * T)
    if option_type == 'call':
        price = discount * ((S - K) * norm.cdf(d) + sigma * np.sqrt(T) * norm.pdf(d))
    elif option_type == 'put':
        price = discount * ((K - S) * norm.cdf(-d) + sigma * np.sqrt(T) * norm.pdf(d))
    return price

def bachelier_digital_cash(S, K, T, r, sigma, Q=1, option_type='call'):
    d = (S - K) / (sigma * np.sqrt(T))
    discount = np.exp(-r * T)
    if option_type == 'call':
        price = Q * discount * norm.cdf(d)
    elif option_type == 'put':
        price = Q * discount * norm.cdf(-d)
    return price

def bachelier_digital_asset(S, K, T, r, sigma, option_type='call'):
    d = (S - K) / (sigma * np.sqrt(T))
    if option_type == 'call':
        price = S * norm.cdf(d)
    elif option_type == 'put':
        price = S * norm.cdf(-d)
    return price
```

$
d_1 = \frac{\ln\left(\frac{F}{K}\right) + \frac{\sigma^2}{2}T}{\sigma \sqrt{T}}, \quad d_2 = d_1 - \sigma \sqrt{T}
$

Vanilla Call Option Price:
$
C = e^{-rT} \left[ FN(d_1) - KN(d_2) \right]

Vanilla Put Option Price:
$
P = e^{-rT} \left[ KN(-d_2) - FN(-d_1) \right]

$
C_{\text{digital}} = Q e^{-rT} N(d_2)
$


Digital Cash-or-Nothing Put Option Price:
$
P_{\text{digital}} = Q e^{-rT} N(-d_2)
$



Digital Asset-or-Nothing Call Option Price:（under risk-neutral valuation）
$
C_{\text{asset}} = F e^{-rT} N(d_1)
$

Digital Asset-or-Nothing Put Option Price:（under risk-neutral valuation）
$
P_{\text{asset}} = F e^{-rT} N(-d_1)
$



```python
def black_model_option(F, K, T, r, sigma, option_type='call'):
    d1 = (np.log(F / K) + 0.5 * sigma**2 * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    discount = np.exp(-r * T)
    if option_type == 'call':
        price = discount * (F * norm.cdf(d1) - K * norm.cdf(d2))
    elif option_type == 'put':
        price = discount * (K * norm.cdf(-d2) - F * norm.cdf(-d1))
    return price

def black_model_digital_cash(F, K, T, r, sigma, Q=1, option_type='call'):
    d2 = (np.log(F / K) - 0.5 * sigma**2 * T) / (sigma * np.sqrt(T))
    discount = np.exp(-r * T)
    if option_type == 'call':
        price = Q * discount * norm.cdf(d2)
    elif option_type == 'put':
        price = Q * discount * norm.cdf(-d2)
    return price

def black_model_digital_asset(F, K, T, r, sigma, option_type='call'):
    d1 = (np.log(F / K) + 0.5 * sigma**2 * T) / (sigma * np.sqrt(T))
    discount = np.exp(-r * T)
    if option_type == 'call':
        price = F * discount * norm.cdf(d1)
    elif option_type == 'put':
        price = F * discount * norm.cdf(-d1)
    return price
```

4.Displaced-Diffusion Model

$
dS_t = (\mu S_t) \, dt + \sigma (S_t + \lambda) \, dW_t
$


Vanilla Call Option Price:
$
C = e^{-rT} \left[ (S_0 + \lambda) N(d_1) - K N(d_2) \right]
$


Vanilla Put Option Price:
$
P = e^{-rT} \left[ K N(-d_2) - (S_0 + \lambda) N(-d_1) \right]
$


Digital Cash-or-Nothing Call Option Price:
$
C_{\text{digital}} = e^{-rT} N(d_2)
$


Digital Cash-or-Nothing Put Option Price:
$
P_{\text{digital}} = e^{-rT} N(-d_2)
$


Digital Asset-or-Nothing Call Option Price:
$
C_{\text{asset}} = (S_0 + \lambda) N(d_1)
$


Digital Asset-or-Nothing Put Option Price:
$
P_{\text{asset}} = (S_0 + \lambda) N(-d_1)
$



```python
def displaced_diffusion_option(S, K, T, r, sigma, beta, option_type='call'):
    S_adj = S + beta
    K_adj = K + beta
    d1 = (np.log(S_adj / K_adj) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    discount = np.exp(-r * T)
    if option_type == 'call':
        price = discount * (S_adj * norm.cdf(d1) - K_adj * norm.cdf(d2))
    elif option_type == 'put':
        price = discount * (K_adj * norm.cdf(-d2) - S_adj * norm.cdf(-d1))
    return price

def displaced_diffusion_digital_cash(S, K, T, r, sigma, beta, Q=1, option_type='call'):
    S_adj = S + beta
    K_adj = K + beta
    d2 = (np.log(S_adj / K_adj) + (r - 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    discount = np.exp(-r * T)
    if option_type == 'call':
        price = Q * discount * norm.cdf(d2)
    elif option_type == 'put':
        price = Q * discount * norm.cdf(-d2)
    return price

def displaced_diffusion_digital_asset(S, K, T, r, sigma, beta, option_type='call'):
    S_adj = S + beta
    K_adj = K + beta
    d1 = (np.log(S_adj / K_adj) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    discount = np.exp(-r * T)
    if option_type == 'call':
        price = S_adj * discount * norm.cdf(d1)
    elif option_type == 'put':
        price = S_adj * discount * norm.cdf(-d1)
    return price
```
